1️⃣ Line Plot (most important)

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [10,20,25,30]

plt.plot(x,y)
plt.title("simple line plot")
plt.xlabel("xaxis")
plt.ylabel("yaxis")

plt.show()
```

2️⃣ Scatter Plot (VERY important for ML)

```python
plt.scatter(x,y)
plt.title("scatter plot")

plt.show()
```

3️⃣ Plot from Pandas (REAL USE)

```python
import pandas as pd

df = pd.read_csv("04Z_ASSET_Workers.csv")

plt.scatter(df["age"], df["salary"])
plt.xlabel("age")
plt.ylabel("salary")
plt.show()
```

4️⃣ Basic Customisation (just enough)

```python
plt.scatter(df["age"], df["salary"])
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid()

plt.show()
```
