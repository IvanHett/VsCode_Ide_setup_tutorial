```python
import matplotlib.pyplot as plt
import numpy as np
xpoints = np.array([2,8])
ypoints = np.array([1,6])

plt.plot(xpoints,ypoints)
plt.show()
```

```python

ypoints = np.array([2,8])

plt.plot(ypoints)
plt.plot(ypoints)
plt.show()
```

Matplotlib does not use variable names (like ypoints) to decide what goes on the y-axis.

Instead, it follows this rule:

plt.plot(x, y) → first argument = x-axis, second = y-axis
plt.plot(y) → only one argument → treated as y, and x is automatically set to [0, 1, 2, ...]
